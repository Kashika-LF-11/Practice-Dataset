```python
pivot_data = data.pivot(index="createdTimestamp", columns="name", values="value")
pivot_data.head()
```

```python
pivot_data.info()
```

```python
pivot_data.describe()
```

```python
pivot_data["gcode.spindle"].unique()
```

```python
pivot_data.index = pd.to_datetime(pivot_data.index)
pivot_data.temperature = pivot_data.temperature.astype('float64')
```

```python
print(pivot_data.index.min(), " , ",  pivot_data.index.max())

print(pivot_data.index.max() - pivot_data.index.min())
```

```python
# Create a basic line plot with time on x-axis and float values on y-axis
plt.figure(figsize=(12, 6))
plt.plot(pivot_data.index, pivot_data['temperature'], marker = 'o', linestyle = '-')  # Line plot with markers
plt.title('Time Series with Float Values on Y-axis')
plt.xlabel('Timestamp')
plt.ylabel('temp')
plt.xticks(rotation=45)  # Rotate x-axis labels for readability
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Assuming 'pivot_data' is already created and has a 'temperature' column

# Calculate statistical attributes of the temperature
stats = pivot_data['temperature'].describe()

# Create a plot
plt.figure(figsize=(12, 6))

# Plot the original temperature values
plt.plot(pivot_data.index, pivot_data['temperature'], marker='o', linestyle='-', label='Temperature')

# Plot statistical values
plt.axhline(stats['mean'], color='blue', linestyle='--', label=f'Mean ({stats["mean"]:.2f})')
plt.axhline(stats['50%'], color='green', linestyle='-.', label=f'Median ({stats["50%"]:.2f})')
plt.axhline(stats['std'], color='orange', linestyle=':', label=f'Standard Deviation ({stats["std"]:.2f})')
plt.axhline(stats['min'], color='red', linestyle='-', label=f'Minimum ({stats["min"]:.2f})')
plt.axhline(stats['25%'], color='purple', linestyle='--', label=f'25th Percentile ({stats["25%"]:.2f})')
plt.axhline(stats['75%'], color='brown', linestyle='--', label=f'75th Percentile ({stats["75%"]:.2f})')
plt.axhline(stats['max'], color='pink', linestyle='-', label=f'Maximum ({stats["max"]:.2f})')

# Add title and labels
plt.title('Temperature Statistics (Mean, Median, Std Dev, Percentiles)')
plt.xlabel('Timestamp')
plt.ylabel('Temperature')

# Rotate x-axis labels for readability
plt.xticks(rotation=45)

# Add grid and legend
plt.grid(True)
plt.legend()

# Show plot
plt.show()
```
